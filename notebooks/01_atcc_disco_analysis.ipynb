```python
import numpy as np
 
# DisCo-level ATC&C loss, full-year 2025 (source: NERC 2025 Annual Report)
discos = ["Kaduna","Jos","Yola","Kano","Benin","Ibadan","Enugu",
          "Port Harcourt","Abuja","Ikeja","Eko"]
discos

atcc_2025 = [71.88,62.15,61.19,44.63,44.29,42.93,40.55,
             39.60,33.96,20.22,16.13]
atcc_2025

#### Plotting  ATCC&C Loss by DisCo, Full-Year 2025 
import matplotlib.pyplot as plt

# Data
discos = ["Eko", "Ikeja", "Abuja", "Port Harcourt", "Enugu", "Ibadan",
          "Benin", "Kano", "Yola", "Jos", "Kaduna"]
losses = [16.1, 20.2, 34.0, 39.6, 40.5, 42.9, 44.3, 44.6, 61.2, 62.1, 71.9]
benchmark = 20.54


# Colour by severity band
def band_color(v):
    if v <= 25:
        return "#2e7d32"   # green
    if v < 40:
        return "#f9a825"   # amber
    if v < 60:
        return "#ef6c00"   # orange
    return "#c62828"       # red


colors = [band_color(v) for v in losses]

fig, ax = plt.subplots(figsize=(11, 6))
bars = ax.bar(discos, losses, color=colors, width=0.62, zorder=3)

# Value labels
for bar, v in zip(bars, losses):
    ax.text(bar.get_x() + bar.get_width() / 2, v + 1, f"{v:.1f}",
            ha="center", va="bottom", fontsize=11)

# Benchmark line
ax.axhline(benchmark, color="#1565c0", linestyle="--", linewidth=2.2,
           label=f"2025 MYTO aggregate benchmark ({benchmark}%)", zorder=4)

# Styling
ax.set_ylabel("ATC&C Loss (%)", fontsize=12)
ax.set_ylim(0, 82)
ax.set_yticks(range(0, 81, 10))
ax.tick_params(axis="y", labelsize=12)
plt.setp(ax.get_xticklabels(), rotation=40, ha="right",
         rotation_mode="anchor", fontsize=11)
ax.grid(axis="y", linestyle=":", color="#cccccc", zorder=0)
ax.spines[["top", "right"]].set_visible(False)
ax.legend(loc="upper left", frameon=False, fontsize=11)

plt.tight_layout()
plt.savefig("atcc_loss_by_disco.png", dpi=150)
plt.show()

# 2. Four-tier performance classification
def classify(v):
    if v <= 25:
        return "Strong"
    elif v <= 40:
        return "Moderate"
    elif v <= 55:
        return "Poor"
    else:
        return "Critical"


bands = {
    d: classify(v) for d, v in zip(discos, atcc_2025)
}

print("\nPERFORMANCE CLASSIFICATION")
for disco, band in bands.items():
    print(f"{disco}: {band}")
    
# 3. Smart Grid Intervention Priority Index (SGIPI)
mn = min(atcc_2025)
mx = max(atcc_2025)

sgipi = {
    d: (v - mn) / (mx - mn)
    for d, v in zip(discos, atcc_2025)
}


def phase(score):
    if score >= 0.70:
        return "Phase 1 - Immediate"
    elif score >= 0.40:
        return "Phase 2 - Short-term"
    else:
        return "Phase 3 - Medium-term"


schedule = {d: phase(s) for d, s in sgipi.items()}
print("\nSMART GRID INTERVENTION PRIORITY INDEX (SGIPI)")
for disco, score in sgipi.items():
    print(f"{disco}: {score:.2f} ({schedule[disco]})")
```
