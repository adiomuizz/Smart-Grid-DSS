```python
import numpy as np
import matplotlib.pyplot as plt

# Data: 2025 quarterly observations
quarters = ["2025/Q1", "2025/Q2", "2025/Q3", "2025/Q4"]

atcc = np.array([39.61, 37.92, 33.27, 34.90])
rev_loss = np.array([200.49, 158.05, 108.75, 139.19])

# Linear regression
slope, intercept = np.polyfit(atcc, rev_loss, 1)
pred = slope * atcc + intercept

# R-squared and Pearson correlation
ss_res = np.sum((rev_loss - pred) ** 2)
ss_tot = np.sum((rev_loss - np.mean(rev_loss)) ** 2)
r2 = 1 - (ss_res / ss_tot)
r = np.corrcoef(atcc, rev_loss)[0, 1]

# Regression line
xs = np.linspace(atcc.min() - 1, atcc.max() + 1, 100)
ys = slope * xs + intercept

# Create figure
fig, ax = plt.subplots(figsize=(9, 6))

ax.scatter(
    atcc, rev_loss,
    s=100,
    zorder=3,
    label="Quarterly observations"
)

ax.plot(
    xs, ys,
    linestyle="--",
    linewidth=2,
    label=f"y = {slope:.2f}x {intercept:+.2f}\n"
          f"R² = {r2:.3f}, r = {r:.3f}"
)

# Label each observation
for q, x, y in zip(quarters, atcc, rev_loss):
    ax.annotate(
        q,
        (x, y),
        textcoords="offset points",
        xytext=(7, 7),
        fontsize=9
    )

# Chart formatting
ax.set_xlabel("Quarterly ATC&C Loss (%)")
ax.set_ylabel("Quarterly Revenue Loss (₦ billion)")

ax.grid(True, linestyle=":", alpha=0.5)
ax.spines[["top", "right"]].set_visible(False)
ax.legend(frameon=False)

plt.tight_layout()
plt.savefig("atcc_vs_revenue_loss.png", dpi=300, bbox_inches="tight")
plt.show()

# Print statistical results
print(f"Regression equation: Revenue Loss = {slope:.3f} × ATC&C Loss {intercept:+.3f}")
print(f"Pearson correlation (r): {r:.3f}")
print(f"Coefficient of determination (R²): {r2:.3f}")
```
